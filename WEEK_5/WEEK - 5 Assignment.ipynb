Q1: What are the key limitations of traditional MapReduce that make Spark a preferred choice for modern big data processing?

Problems with Traditional MapReduce
- It writes intermediate results to disk after every processing stage.
- It has a complex programming model and mainly supports batch processing.

Why Spark is preferred
- Spark uses in-memory computing, so it processes data much faster.
- It supports SQL, streaming, and machine learning within a single framework
- It is easier to use and write applications.

Q2: Explain how Spark uses In-Memory Computing to speed up iterative machine learning algorithms compared to disk-based systems.

- Spark stores data in memory (RAM) instead of writing it to disk after every step.
- Machine learning algorithms use the same data many times.
- Since the data is already in memory, Spark does not need to read it from disk again and again.
- This reduces processing time and makes machine learning algorithms much faster than disk-based systems like MapReduce.

Q10: Write the code to revise a column named raw_timestamp by casting it to a TimestampType and renaming it to event_time.``

```python
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType
spark_df = spark_df.withColumn("event_time", col("raw_timestamp").cast(TimestampType())).drop("raw_timestamp")
spark_df.show()
```
